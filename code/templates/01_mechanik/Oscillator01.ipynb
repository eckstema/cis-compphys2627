## Oscillator 01

Solution of equation $ \ddot x = - \sin(x) $ with 

(1) forward Euler

(2) Runge Kutta 4

(3) Verlet Störmer

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# force as function of x:
def f(x):
    return -x
    #return -np.sin(x)

def energy(x,v):
    return 0.5*v*v-np.cos(x)

```

```python
# update variables (x_n,v_n) to (x_{n+1},v_{n+1}) using forward Euler 
def euler_step(xn,vn,h):
    xn1 = xn + h*vn
    vn1 = vn + h*f(xn)
    return xn1,vn1


def euler_full(h,nmax,x0,v0):
    xn=x0
    vn=v0
    xpoints = np.zeros(nmax)
    vpoints = np.zeros(nmax)
    tpoints = np.zeros(nmax)
    for n in range(nmax):
        tn=n*h
        tpoints[n]=tn
        xpoints[n]=xn
        vpoints[n]=vn
        xn,vn=euler_step(xn,vn,h)
    return tpoints,xpoints,vpoints
```

```python
fig, ax = plt.subplots() 

tmax=2.0*np.pi*4
nmax=200
h=tmax/(nmax-1)
x0=0
v0=1
tau=2*np.pi
t1,x1,v1=euler_full(h,nmax,x0,v0)
t2,x2,v2=euler_full(h/10,10*nmax,x0,v0)
t3,x3,v3=euler_full(h/100,100*nmax,x0,v0)


#plt.plot(xpoints, vpoints)
ax.set_xlabel('time*Omega')
ax.set_ylabel('x')
#ax.plot(t1,energy(x1, v1),'r-',label='euler, h=0.1')
ax.plot(t1/tau,x1,'b-',label=f'euler, h={h:.3f}')
ax.plot(t2/tau,x2,'r-',label=f'euler, h={h/10:.3f}')
ax.plot(t3/tau,x3,'g-',label=f'euler, h={h/100:.3f}')
ax.plot(t2/tau,np.sin(t2),'k-',label='exact')
ax.legend()
plt.show()

```

```python
def rk4_step(xn,vn,h):
    k1x = vn
    k1v = f(xn)
    k2x = vn+h*k1v/2
    k2v = f(xn+h*k1x/2)
    k3x = vn+h*k2v/2
    k3v = f(xn+h*k2x/2)
    k4x = vn+h*k3v
    k4v = f(xn+h*k3x)
    kx = (k1x+2*k2x+2*k3x+k4x)/6
    kv = (k1v+2*k2v+2*k3v+k4v)/6
    xn = xn + h*kx
    vn = vn + h*kv
    return xn,vn

def rk4_full(h,nmax,x0,v0):
    xn=x0
    vn=v0
    xpoints = np.zeros(nmax)
    vpoints = np.zeros(nmax)
    tpoints = np.zeros(nmax)
    for n in range(nmax):
        tn=n*h
        tpoints[n]=tn
        xpoints[n]=xn
        vpoints[n]=vn
        xn,vn=rk4_step(xn,vn,h)
    return tpoints,xpoints,vpoints

```

```python
#Verglech RKe - vs Euler

tmax=2.0*np.pi*40
nmax=20000
h=tmax/(nmax-1)
x0=0
v0=1
tau=2*np.pi


fig, ax = plt.subplots() 

t1,x1,v1=euler_full(h,nmax,x0,v0)
t2,x2,v2=rk4_full(h*20,int(nmax/20),x0,v0)


#plt.plot(xpoints, vpoints)
ax.set_xlabel('time*Omega')
ax.set_ylabel('x')
#ax.plot(t1,energy(x1, v1),'r-',label='euler, h=0.1')
#ax.plot(t1/tau,x1,'b-',label=f'euler, h={h:.3f}')
ax.plot(t1/tau,x1,'r-',label=f'euler, h={h:.3f}')
ax.plot(t2/tau,x2,'g-',label=f'RK4, h={h*20:.3f}')
ax.plot(t2/tau,np.sin(t2),'k--',label='exact')
ax.legend()
plt.show()
```

```python
def verlet_step(xn,vn,vn12,h):
    xn = xn+ vn12*h
    fn=f(xn);
    vn=vn12+0.5*h*fn
    vn12 = vn12 + h*fn
    return xn,vn,vn12

def verlet_full(h,nmax,x0,v0):
    xn=x0
    vn=v0
    fn=f(xn);
    vn12=v0+0.5*h*fn; 
    xpoints = np.zeros(nmax)
    vpoints = np.zeros(nmax)
    tpoints = np.zeros(nmax)
    for n in range(nmax):
        tn=n*h
        tpoints[n]=tn
        xpoints[n]=xn
        vpoints[n]=vn
        xn,vn,vn12=verlet_step(xn,vn,vn12,h)
    return tpoints,xpoints,vpoints

```

```python
# test der Energieerhaltung RK4 vs Verlet


tmax=2.0*np.pi*4000
nmax=200000
h=tmax/(nmax-1)
x0=0
v0=1
tau=2*np.pi


fig, ax = plt.subplots() 

t1,x1,v1=rk4_full(h,int(nmax),x0,v0)
t2,x2,v2=verlet_full(h,int(nmax),x0,v0)


#plt.plot(xpoints, vpoints)
ax.set_xlabel('time*Omega')
ax.set_ylabel('|E(t)-E(0)|')
#ax.plot(t1,energy(x1, v1),'r-',label='euler, h=0.1')
#ax.plot(t1/tau,x1,'b-',label=f'euler, h={h:.3f}')
#ax.plot(t1/tau,np.abs(0.5*v1**2+0.5*x1**2-0.5),'r-',label=f'euler, h={h:.3f}')
ax.plot(t2/tau,0.5*v2**2+0.5*x2**2-0.5,'g-',label=f'verlet, h={h:.3f}')
ax.plot(t1/tau,0.5*v1**2+0.5*x1**2-0.5,'r-',label=f'RK4, h={h:.3f}')
#ax.plot(t2/tau,np.ones_like(t2),'k--',label='exact')
ax.legend()
#ax.set_yscale('log')
plt.show()
```
