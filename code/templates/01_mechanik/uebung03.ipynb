## Übung 03 – Schwingungen einer linearen Federkette

Wir betrachten eine Kette von $N+1$ Massen, die durch Federn der Federkonstanten $k$ gekoppelt sind. Zusätzlich sind die beiden Enden der Kette an festen Punkten befestigt. Die Ruhelage der Masse $j$ sei $aj$ ($j=0,\ldots,N$), und $q_j$ bezeichne die Auslenkung der Masse $j$ aus der Ruhelage.

Sie können Teile des Programms `CoupledPendulum.ipynb` als Ausgangspunkt verwenden.

**Einheiten:** Wir setzen $a=1$, $m=1$ und $k=1$. Längen werden damit relativ zu $a$ und Zeiten in Einheiten von $\sqrt{m/k}$ gemessen.

Die folgende Zelle zeichnet eine schematische Darstellung der Kette. Die grauen Punkte sind fest, die blauen Punkte sind die beweglichen Massen.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Schematische Darstellung; die Anzahl dient nur der Illustration.
n_draw = 8
x = np.arange(n_draw + 2)
y = np.zeros_like(x, dtype=float)

fig, ax = plt.subplots(figsize=(9, 2.0))
for left in range(n_draw + 1):
    xs = np.linspace(x[left], x[left + 1], 50)
    ys = 0.10*np.sin(10*np.pi*(xs - x[left]))
    ax.plot(xs, ys, color="0.35", lw=1.2)
ax.scatter(x[[0, -1]], y[[0, -1]], marker="s", s=55, color="0.35", label="fest")
ax.scatter(x[1:-1], y[1:-1], s=70, color="C0", zorder=3, label="Massen")
for j, xpos in enumerate(x[1:-1]):
    ax.text(xpos, 0.22, rf"$j={j}$", ha="center", fontsize=9)
ax.set(xlim=(-0.4, n_draw + 1.4), ylim=(-0.35, 0.45))
ax.set_aspect("equal")
ax.axis("off")
ax.legend(loc="lower center", ncol=2, frameon=False)
plt.show()
```
