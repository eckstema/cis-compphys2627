# Introduction: plotting a function

Run the cell below to test your Computational Physics environment.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 400)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y, label=r"$\sin(x)$")
ax.set(xlabel="x", ylabel="f(x)", title="Simple function (sin)   ")
ax.grid(True)
ax.legend()
plt.show()
```
